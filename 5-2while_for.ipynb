총 5명의 학생이 시험을 보았는데 시험 점수가 60점 이상이면 합격이고 그렇지 않으면 불합격이다. 합격인지, 불합격인지 결과를 보여 주시오.

```python
marks = [90,25,67,45,80]

number = 0
for mark in marks :
    number += 1
    if mark>= 60 : 
        print("%d 번 학생은 합격입니다." %number)
    else :
        print("%d 번 학생은 불합격입니다." %number)
```

```python
marks = [90,25,67,45,80]

number = 0
for mark in marks :
    number += 1
    if mark > 60 :
        continue
    print ("%d번 학생은 불합격입니다 화이팅하세요" %number)
```

### 리스트 컴프리헨션 (list comprehension)

## 리스트 컴프리헨션 문법
[표현식 for 항목 in 반복_가능_객체 if 조건문]

```python
# 리스트 컴프리헨션안에 if 문 추가하기
result = [num*2 for num in a if num%2 ==0]
print(result)
```

```python
#리스트 컴프리헨션안에 if 문 추가하기
result = [x*y for x in range (2,10)
               for y in range (1,10)]
print(result)
```

```python
for prediction in [True, False]:

    # 맞히면 문을 통과할 수 있으므로 문은 열린다.
    if prediction:
        actual_door = True

    else:
        # "문이 안 열린다"고 예측했다고 가정
        # 만약 그 예측이 맞다면 통과해야 하므로
        actual_door = True

    if prediction == actual_door:
        print(prediction, "→ 가능한 답")
    else:
        print(prediction, "→ 모순")
```
