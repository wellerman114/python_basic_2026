## 문자열 포맷 코드
- %s : 문자열 (string)
- %c : 문자 1개 (character)
- %d : 정수 (inteher)
- %f : 부동소수 (floating-point)
- %o : 8진수
- %x : 16진수
- %% : Literal % (문자 '% 자체')

```python
#1) %는 숫자도 문자열로 대입
"i have %s apples" % 3
```

```python
"pi is %s" % 3.141592 
```

```python
# 2) 포매 코드와 문자 %를 같이 사용하려면 %%
"Error is %d%%." % 98
```

```python
# 3) 포맷 코드와 숫자 함게 사용하기
 #정렬과 공백

"%10s" % "hi"

"%-10sJane" % 'hi'

 #소수점 표현하기
"%0.4f" % 3.42134234
```

```python
"%10.4f" % 3.42134234
```

## format() 함수를 사용한 포맷팅

```python
# 1) 숫자 바로 대입 
"i eat {} apples" .format(3)
```

```python
# 2) 문자열 바로 대입
"i eat {} apples" .format("five")
```

```python
# 3) 숫자 값을 나타내는 변수로 대입
number = 3
"i eat {} apples" .format(number)
```

```python
# 4) 2개 이상의 값 넣기
number = 10
day = "three"
"i ate %d apples. so I was sick for %s days." % (number, day)
```

```python
# 5) 이름으로 넣기
"i ate {number} apples. so i was sick for {day} days." .format(number=10, day=3)
```

```python
# 6) 인덱스와 이름을 혼용해서 넣기
"i ate {0} apples. so i was sick for {day} days." .format(10,day=3)
```

```python
# 7~8) 왼쪽 정렬과 오른쪽 정렬
"{:>10}" .format("hi")
```

```python
"{:<10}" .format("hi")
```

```python
# 9) 가운데 정렬
"{:^10}" .format("hi")
```

```python
# 10) 공백 채우기
"{:ㅗ<10}" .format("hi")
```

```python
"{:ㅜ>10}" .format("hi")
```

```python
"{:ㅗ^10}" .format("hi")
```

```python
# 11) 소수점 표현
e = 2.7182818284590452353602874713527
"{:.6f}".format(e)
```

```python
pi = 3.1415926535897932384626433832795
"{:.20f}" .format(pi)
```

```python
# 12) {또는} 문자 표현
"{{and}}" .format()
```

# 문자열 관련 함수들
- 문자열 자료형은 자체적으로 함수를 가지고 있다. 이 함수를 '문장려 내장 함수'라고 한다. 내장 함수를 사용하려면 문자열 변수 이름 뒤에 '.'를 붙인 후 함수 이름을 써 주면 된다.

```python
# count 특정 문자의 갯수새기
a = "hobby"
a.count('b')
```

```python
# find 위치 찾기 1
a = "Python is the best choice"
a.find('b')
```

```python
a.find('k') # 찾는 문자가 없으면 -1 반환 find는 예외처리 가능
```

```python
# index 위치 찾기 2
a = "Life is too short"
a.index('t')    
```

```python
a.index('k') # 찾는 문자가 없으면 에러발생 index는 예외처리 불가
```

```python
#join 문자열 삽입
",".join('abcd')     
```

```python
",".join(['a','b','c','d'])
```

```python
#upper 소문자를 대문자로 바꾸기
a = "hi"
a.upper()
```

```python
# lower 대문자를 소문자로 바꾸기
a = "HI" 
a.lower()
```

```python
#lstrip 왼쪽 공백 지우기
a = " hi "
a.lstrip()
```

```python
#rstrip 오른쪽 공백 지우기
a = " hi "
a.rstrip()
```

```python
#strip 양쪽 공백 지우기
a = "   hi   "
a.strip()
```

```python
# 문자열 바꾸기
a = "Life is too short"
a.replace("Life", "Your leg")
```

```python
# split 문자열 나누기
a = "Life is too short"
a.split()
```

```python
b = "a:b:c:d"
b.split(':')
```

```python
#착각하기 쉬운 문자열 함수
a = 'hello'
a.upper()
```

```python
a   #원본 데이터는 변경 x
```

```python
a = a.upper()
a     #변경된 데이터는 변수에 대입해야 변수 값이 변경됨.
```

왜냐하면 a.upper()를 실행하면 upper 함수는 a 변수의 값 자체를 변경하는 것이 아니라 대문자로 바꾼 값을 리턴하기 때문이다. 문자열은 자체의 값을 변경할 수 없는 immutable 자료형이다. 따라서 a 값을 'HI' 로 바꾸고 싶다면 다음과 같이 대입문을 사용해야 한다.

** upper 뿐만 아니라 lower, join, lstrip, rstrip, strip, replace, split 함수는 모두 이와 같은 규칙이 적용되어 문자열 자체의 값이 변경되는 것이 아니라 변경된 값을 리턴한다는 사실에 주의하자.
