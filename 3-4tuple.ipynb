
# 튜플의 특징
#### 튜플(tuple)은 몇 가지 점을 제외하곤 리스트와 거의 비슷하며 리스트와 다른 점은 다음과 같다.
- 리스트는[], 튜플은()으로 둘러싼다.
- 리스트는 요솟값의 생성, 삭제, 수정이 가능하지만, 튜플은 요솟갑을 바꿀 수 없다.

## 1. 튜플은 어떻게 만들까?

```python
t1=()
t2 = (1,) #끝에 쉼표
t3 = (1,2,3) #여러개면 끝에 쉼표 생략 가능
t4 = 1,2,3, #t4는 괄호 생략 가능
t5 = ('a','b',('ab','cd')) 
#튜플과 리스트의 가장 큰 차이는 요솟갑을 변화시킬 수 있는지으 여부 
```

## 2. 튜플의 요솟갑을 삭제 변경하려고 하면 어떻게 될까?

```python
# 튜플 요솟갑을 삭제하려 할 때
t1 = (1,2,'b''a')
del t1[0]
```

```python
#변경 시도시
t1 = (1, 2, 'a', 'b')

t1[0] = 'c'
```
