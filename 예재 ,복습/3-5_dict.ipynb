## 3. 딕셔너리 쌍 추가, 삭제하기

```python
a = {1: 'a'}

a[2] = 'b'
a
```

```python
a['name'] = 'pey'
a
```

```python
a[3] = [1,2,3]
a
```

```python
# del 키워드로 요소 삭제하기
del a[1]
a
```

```python
del a['name']
a
```

## 5. 딕셔너리 관련 함수

key 리스트 만들기 : keys()

```python
a = {'name': 'ajw', 'phone': '010-1234-1234', 'birth': '1118'}
```

```python
a.keys()
```

```python
type(a.keys())
```

```python
list(a.keys())
```

```python
type(list(a.keys()))
```

```python
for k in a.keys():
    print(k)
```

value 리스트 만들기 : values()

```python
a.values()
```

key, value 쌍 얻기 : items()

```python
a.items()
```

key: value 쌍 모두 지우기 : clear()

```python
a.clear()
a
```

key로 value얻기 : get()

```python
a = {'name': 'ajw', 'phone': '010-1234-1234', 'birth': '1118'}

a.get('phone')
```

```python
print(a.get('age'))
```

```python
print(a['age'])
```

```python
# 딕셔너리 안에 찾는 key가 없으면 디폴트 값을 가져오게 할 수 있다.
# get(key, 'default value')

a.get('age', 25)
```

```python
a
```

```python
a
```
