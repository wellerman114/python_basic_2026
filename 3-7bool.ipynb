## 자료형의 참과 거짓
- 값이 비어 있으면 거짓 ("".[],{}) 
- 비어있지 않으면 참 쉽게 말해 0,1로 구분하면됨 
- none 은 거짓을 뜻함

```python
#while ~~하는 동안에   (복습 pop은 뒤에서부터 끄집어 내기 그리고 삭제)
a = [1,2,3,4]
while a :
    print(a.pop())
```

```python
print(a)
```

```python
a==True
```

```python
# if (이라면) else (아니면)
if[]:
    print("참")
else:
    print("거짓")
```
