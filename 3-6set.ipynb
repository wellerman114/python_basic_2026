## 3. 교집합 합집합 차집합 구하기

```python

a1 = set([1,2,3,4,5,6])
a2 = set([4,5,6,7,8,9])
```

```python
#교집합 구하기 (각 세트 사이에 & 사용) (intersection 함수 사용도 가능)
a1 & a2
```

```python
a1.intersection(a2)
```

```python
#합집합 구하기 (|) (union 함수 사용)
a1 | a2
```

```python
a1.union(a2)
```

```python
# 차집합 구하기 (-를 사용하여 구하기) (diffenence 함수 사용)
a1-a2
```

```python
a2-a1
```

```python
a1.difference(a2)
```

```python
a2.difference(a1)
```

## 4. 집합 자료형 관련 함수
- add() 값 1개 추가하기
- update() 값 여러개 추가하기
- remove() 특정 값 제거하기

```python
#add()
a1 = set([1,2,3])
a1.add(4)
a1
```

```python
#update([]) 소괄호 안에 대괄호 넣어야 함
a1.update([5,6,7])
a1
```

```python
#remove() 특정 값 제거하기
a1.remove(5)
a1
```
