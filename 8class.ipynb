# 1. 클래스가 필요한 이유

```python
# 계산기 만들어보기

result = 0

def add(num):
    global result
    result += num
    return result

print(add(3))
print(add(4))
```

```python
# 계산기 2대가 필요한 상황
result1 = 0
result2 = 0

def add1(num):
    global result1

    result1 += num

    return result1

def add2(num):  # 계산기2​

    global result2

    result2 += num

    return result2

print(add1(3))

print(add1(4))

print(add2(3))

print(add2(7))
```

```python
# 클래스로 해결

class Calculator: # 생성자 함수 (클래스 변수 초기화)
    def __init__(self): 
        self.result = 0

    def add(self,num):
        self.result += num
        return self.result

    def sub(self,num):
        self.result -= num
        return self.result

```

# 5. 클래스의 상속

```python
class FourCal:
    #생성자 함수

    def __init__(self,first,second):
        self.first = first
        self.second = second

        
    #숫자 2개 입력받는 함수(매서드)
    def setdata(self,first,second):
        self.first = first
        self.second = second

    #더하기 함수
    def add(self):
        result = self.first + self.second
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def mul(self):
        result = self.first * self.second
        return result

    def div(self):
        result = self.first / self.second
        return result

    
```

```python
class MoreFourCal(FourCal):
    def pow(self):
        result = self.first ** self.second
        return result

```

```python
a = MoreFourCal(15,17)
a.pow()
```

# 6. 매서드 오버라이딩(재정의)

```python
class SafeFourCal(FourCal):
    def div(self):
        if self.second == 0:
            return 0
        else:
            result = self.first / self.second
            return result

        
```

```python
a = SafeFourCal(15,0)
a.div()
```
