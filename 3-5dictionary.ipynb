# 딕셔너리

## 1. 딕셔너리란?
딕셔너리는 단어 그대로 '사전' 이라는 뜻
key와 value를 한쌍으로 가지는 자료형
딕셔너리는 리스트나 튜플처럼 순차적으로 (sequential) 해당 요솟값을 구하지 않고 key와 value를 얻는다. 이것이 바로 딕셔너리의 가장 큰 특징이다.

## 2. 딕셔너리는 어떻게 만들까?

# 딕셔너리의 기본형 
{key : value,key : value,key : value}

```python
dic = {'name':'pey', 'phone':'010-9999-1234','birth':'1118'}
dic
```

```python
dic['name']
```

```python
# key값을 정수로 써보기
a = {1:'hi',2:'hello'}
a
```

```python
# value는 리스트도 넣을 수 있다.
a = {'a':[1,2,3]}
a['a']
```

## 4. 딕셔너리를 사용하는 방법

```python
kdk = {'name':'kdk','age':'27','weight':'76' }
kdk
```

```python
grade ={'pey':10,'julliet':99}
grade['pey']
```

```python
grade['julliet']
```

```python
# 딕셔너리를 만들 때 주의사항
a = {1:'a',1:'b'}
a
#딕셔너리에서 key는 고유한 값이므로 중복되는 key값을 설정 해 놓으면 하나를 제외한 나머지 것들이 모두 무시 된다는 점을 주의해야 한다.
```

```python
a = {[1,2]:'hi'}
a
#key에 리스트는 쓸 수 없다 
```

```python
a = {(1,2):'hi'}
a
#하지만 튜플은 key로 쓸 수 있다.
```

## 5. 딕셔너리 관련함수

```python
#keys() 키 리스트 만들기
a={'name':'pey','phone':'111111111','birth':1118}
a.keys()
```

```python
# 밸류 리스트 만들기
a.values()
```

```python
#items()키 밸류 쌍 얻기
a.items()
```

```python
print(a)
```

```python
a.clear()
a
```

```python
#get key로value 얻기
a = {'name': 'pey', 'phone': '010-9999-1234', 'birth': '1118'}
a.get('name')
```

```python
a.get('phone')
```

```python
#in 해당 key가 딕셔너리 안에 있는지 조사하기
a = {'name':'pey', 'phone':'010-9999-1234', 'birth': '1118'}
'name' in a
```
