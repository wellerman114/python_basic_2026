## 1. 예외 처리 기법


try - except 구문

```python
try:
    4/0

except ZeroDivisionError as e:
    print(e)
```

try-fanally 구문

```python
try:
    f = open('foo.txt','w')

finally:
    f.close()
```

여러개 예외 처리

```python
try :
    a = [1,2]
    print(a[3])
    4/0

except ZeroDivisionError:
    print("0으로 못나눔")
except IndexError:
    print("인덱싱 불가")
```

try - else 구문 


```python
try:
    age=int(input('나이를 입력해라:'))
except:
    print('입력이 정확하지 않습니다.')

else:
    if age <= 18:
        print("미성년자는 출입금지입니다")
    else:
        print("환영합니다")
```

## 3. 예외 발생 
- raise 명령어를 사용해 예외를 강제로 발생

```python
class Bird:
    def fly(self):
        raise NotImplementedError
```

```python
class Eagle(Bird):
    pass

eagle = Eagle()
eagle.fly()

#Bird클래스의 fly매서드가 수행되어 NotImplemintError가 발생
```

```python
# 매서드 오버라이딩 (이글 클래스에 플라이 함수를 구현)
class Eagle(Bird):
    def fly(self):
        print("very fast")

eagle = Eagle()
eagle.fly()
```

4. 예외 만들기

```python
# 내장 클래스인 Exception클래스를 상속
class MyError(Exception):
    pass
```

```python
#별명을 출력하는 함수
def say_nick(nick):
    if nick == 'fool':
        raise MyError()
    print(nick)
```

```python
#say_nick 함수호출
say_nick("angel")
```

```python
say_nick("fool")
```

```python
try:
    say_nick('angel')
    say_nick('fool')
except MyError:
    print('허용되지 않는 별명입니다')
```

```python
#오류 메시지를 사용하고 싶다면
try:
    say_nick('angel')
    say_nick('fool')
except MyError as e:
    print(e)
```

```python
#오류 메시지를 보이게 하려면
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명임"
```

```python
#오류 메시지 출력 
try:
    say_nick('angel')
    say_nick('fool')
except MyError as e:
    print(e)
```
