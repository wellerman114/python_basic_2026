2-2 파일을 쓰기 모드로 열어 내용쓰기

```python
f = open('newfile.txt','w')

for i in range(1,11):
    data = f'{i}번째 줄입니다.\n'
    f.write(data)

f.close()
```

2-3 파일을 읽는 여러가지 방법

```python
#한줄읽기 : readline()
f = open('newfile.txt','r')

line = f.readline()
print(line)

f.close()
```

```python
f = open('newfile.txt','r')

while True:
    line = f.readline()
    if not line : break
    print(line)

f.close()
```

```python
# 여러줄 읽기 : readlines()
f = open('newfile.txt','r')

lines = f.readlines()
print(lines)

for line in lines:
    print(line)

f.close()
```

```python
# 줄바꿈 (\n) 문자 제거하기 .strip()
f = open('newfile.txt','r')

lines = f.readlines()


for line in lines:
    print(line.strip())

f.close()
```

```python
# 파일 전체 내용 읽기 : read()
f = open('newfile.txt','r')

data = f.read()
print(data)

f.close()
```

```python
# 파일 객체를 for문과 함깨 사용하기
f = open('newfile.txt','r')
for line in f :
    print(line.strip())

f.close()
```
